Run four editable workflows through the existing CAS language. The 17 mathematical expectations were fixed independently before execution. This notebook contains no captured output and does not by itself establish V1 acceptance. See [the complete run instructions and derivations](../examples/kernel-v1.md).

## Prepare the existing execution environment

Use the **CasDsl (Lean 4)** kernel configured for this checkout, with the built `CasDsl.Notebook` prelude. From the repository root, set `CAS_LEAF_CONTRACT_PATH` to the contract checkout and `CAS_SAGE_PYTHON` to an installed Python executable with Sage. Run `python3 scripts/run_kernel_v1_demo.py --prepare-only` before starting Jupyter with that environment.

The command below uses the default `../kernel-v1-run/` output directory. If you selected other paths, replace it with the absolute-path command printed by `--prepare-only`. The selected configuration consists of the diagnostic registrations plus only the existing set pullback and cardinality registrations. No implementation priority is added.

## The dependent computations

| Workflow source | Retained values and expected observations |
| --- | --- |
| `01-selected-arithmetic.cas` | `17-29 → -12 → 144 → 144/35 → 99/35 → 3`; the selected modular chain gives `3 → 4 → 1` in `ℤ/5`. |
| `02-finite-pullback.cas` | The pullback of `[0,1,1]` and `[1,0]` has three elements; its retained underlying set is reused in a product of cardinality nine. |
| `03-matrix-iteration.cas` | One rational matrix sends the retained vectors successively to `(-3,8,8)`, `(5,32,-6)`, and `(75,14,-28)`. |
| `04-presentation-transport.cas` | `1+2x → 2+2y → y+2 → x` uses forward transport, multiplication, and inverse transport. |

Each file keeps its own `let` scope and computational data for later operations. `computing` requests the realization path and records actual assertion dispatches. The output reports which assertions hold and details any other outcome.

## Edit a connected chain

The source files are in `examples/kernel-v1/`. For experimentation, copy a file, edit it, and point `#cas_tests` at that file; the command-line runner also accepts `--suite /path/to/my-experiment.cas`. Retain the independently specified originals and derive new expected values before consulting a backend answer.

Keep dependent expressions in one `.cas` file with blank lines between items. A standalone `#cas "..."` starts from an empty CAS scope, so a `let` in one such command does **not** persist into the next notebook cell. Different `.cas` files have separate scopes. This notebook uses the existing file harness to preserve dependencies; it adds no notebook state machinery.

For a copied file in this notebook, use `--prepare-only --suite /path/to/my-experiment.cas` and paste the printed command, including its matching inventory. Use the same `--suite` when checking that report.
