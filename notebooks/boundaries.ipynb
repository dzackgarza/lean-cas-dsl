# CasDsl — the refusal catalogue

Every refusal in CasDsl is a sentence: it names the operation, the
object, and the reason. This notebook demonstrates them live — each
output below is the genuine kernel refusal — so that
[demo.ipynb](demo.ipynb) can stay a runnable trail.

A caveat for interactive use: the document model refuses to build
on a failed cell, so running top-down stops at the first refusal —
the stop is itself the design. To exercise a later refusal live,
comment out the refusal cells above it (`--` begins a comment,
exactly as in Lean) and re-run.

## Reserved constants

`e` and `i` are never shadowable — `e^t` and `2 + 2i` mean the same
thing in every session state, so a rebinding refuses:

```lean4
let e := 5 in ℤ
```

```lean4
let i := 3 in ℤ
```

## Cross-category equality

`F` is a module and `{0, 1, 2, 3}` a set: bare `=` never inserts
the forgetful functor, so the equality is false. The Sets question
is `set_eq`, receiver transported — and it answers:

```lean4
let F := ℤ/4 in SmallModules(ℤ)
```

```lean4
assert F = {0, 1, 2, 3}
```

```lean4
F.set_eq({0, 1, 2, 3})
```

## Symbolic bodies are never evaluated at a point

A body the polynomial engine cannot express is symbolic: the
expression is presented for a limit, a definite integral or a
Taylor expansion, and evaluation at a point is not implemented:

```lean4
let sine: ℝ → ℝ := t ↦ sin(t)
let recip := t ↦ 1/t in ℝ → ℝ
```

```lean4
sine(0)
```

```lean4
recip(2)
```

## A function needs its domain and codomain

A bare `↦` has no ascription to check against:

```lean4
let bad := t ↦ t^2
```

## Domains must compose, and arguments must belong

```lean4
let h := t ↦ t² + 1 in ℝ → ℝ
let m2: ℕ → ℕ := n ↦ 2n
```

```lean4
assert (h ∘ m2)(t) = t
```

```lean4
m2(-1)
```

## A morphism is not an object

Ascribing a hom to `Mod(ℚ)` states membership among the category's
objects — and a map is not one:

```lean4
let gm: QQ^3 -> QQ := (a, b, c) |-> a + b - c in Mod(QQ)
```

## Reserved spellings

Spellings SPEC.md fixes for features not implemented yet still
parse, and refuse by name — never as syntax errors:

```lean4
let Rf := CC[x_0, x_1, ..., x_9]
```

```lean4
AA^2(QQ)
```

## `E[[t]]/(t^k)` is a truncation request, not a ring here

It occurs only as the target of `map … to`; membership in it is not
a question this system answers:

```lean4
let sq(t) = ∑_{n ∈ ℕ} n^2 t^n ∈ ℤ[[t]]
```

```lean4
assert sq ∈ ℤ[[t]]/(t^5)
```

## NoImplementation — availability without an implementation

`det` is declared on Mat₂(ℤ/5); no route executes it there yet. The
refusal is a structured gap, never a narrowing of the mathematics:

```lean4
let N := [1, 2; 3, 4] in Mat₂(ℤ/5)
```

```lean4
N.det()
```
