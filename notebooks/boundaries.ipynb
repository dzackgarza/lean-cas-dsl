# CasDsl — the refusal catalogue

Every refusal in CasDsl is a sentence: it names the operation, the
object, and the reason. This notebook collects the deliberate
failure modes so that [demo.ipynb](demo.ipynb) stays a runnable
trail — the document model refuses to build on a failed cell, so a
live refusal blocks every cell below it.

**How to use this notebook.** Every offending line ships commented
(`--` begins a comment, exactly as in Lean). Uncomment one line,
run the cell, read the refusal, and re-comment it before moving
on — while it stays failed, the cells below are blocked, which is
the document model doing its job. Each refusal's wording is quoted
below its section, and the wordings are pinned by the test suite.

## Reserved constants

`e` and `i` are never shadowable — `e^t` and `2 + 2i` mean the same
thing in every session state.

> \`e\` is reserved for Euler's constant (\`e^t\` = exp(t)) and
> cannot be rebound

> \`i\` is reserved for the imaginary unit i ∈ ℂ, so it cannot be
> rebound

```lean4
assert 2 + 2i ∈ ℂ
-- let e := 5 in ℤ
-- let i := 3 in ℤ
```

## Cross-category equality

`F` is a module and `{0, 1, 2, 3}` a set: bare `=` never inserts
the forgetful functor, so the equality is false. The Sets question
is `F.set_eq({0, 1, 2, 3})`, receiver transported.

> assertion is false: F = {0, 1, 2, 3}

```lean4
let F := ℤ/4 in SmallModules(ℤ)
-- assert F = {0, 1, 2, 3}
```

## Symbolic bodies are never evaluated at a point

A body the polynomial engine cannot express is symbolic: the
expression is presented for a limit, a definite integral or a
Taylor expansion.

> t ↦ sin(t) has a symbolic body: the expression is presented for a
> limit, a definite integral or a Taylor expansion, and evaluation
> at a point is not implemented for symbolic bodies

```lean4
let sine: ℝ → ℝ := t ↦ sin(t)
let recip := t ↦ 1/t in ℝ → ℝ
-- sine(0)
-- recip(2)
```

## A function needs its domain and codomain

> \`t ↦ …\` is a function definition: bind it with its domain and
> codomain, as in \`let h := t ↦ t^2 + 1 in ℝ → ℝ\`

```lean4
let good := t ↦ t^2 in ℝ → ℝ
-- let bad := t ↦ t^2
```

## Domains must compose, and arguments must belong

> ℕ → ℕ and ℝ → ℝ do not compose: the target of the right factor is
> not the source of the left

> -1 is not an element of ℕ

```lean4
let h := t ↦ t² + 1 in ℝ → ℝ
let m2: ℕ → ℕ := n ↦ 2n
-- assert (h ∘ m2)(t) = t
-- m2(-1)
```

## A morphism is not an object

> a hom is a morphism, not an object of Mod(ℚ); ascribing it there
> is refused rather than read as membership — the arrow already
> names its domain and codomain

```lean4
let gφ: ℚ³ → ℚ := (a, b, c) ↦ a + b - c
-- let gm: QQ^3 -> QQ := (a, b, c) |-> a + b - c in Mod(QQ)
```

## Reserved spellings

Spellings SPEC.md fixes for features not implemented yet still
parse, and refuse by name — never as syntax errors.

> D[x_0, x_1, ..., x_n] — a polynomial algebra on a family of
> indeterminates (SPEC.md §Ellipses) — is not implemented: the
> spelling is reserved for it. One indeterminate — \`ℂ[x]\` — is
> available

> AA^n(K) — affine n-space over K — is not implemented: the
> spelling is reserved for it

```lean4
ℂ[x]
-- let Rf := CC[x_0, x_1, ..., x_9]
-- AA^2(QQ)
```

## `E[[t]]/(t^k)` is a truncation request, not a ring here

> \`E[[t]]/(t^5)\` occurs here only as the target of
> \`map … to\` — a request that a series be presented by its first
> 5 coefficients. The quotient ring itself is not presented in
> CasDsl, so it answers no membership, cardinality or inclusion
> question

```lean4
let sq(t) = ∑_{n ∈ ℕ} n^2 t^n ∈ ℤ[[t]]
-- assert sq ∈ ℤ[[t]]/(t^5)
```

## NoImplementation — availability without an implementation

`det` is declared on Mat₂(ℤ/5); no route executes it there yet. The
refusal is a structured gap, never a narrowing of the mathematics.

> NoImplementation: 'det' is mathematically available for
> x = [1, 2; 3, 4] ∈ Mat₂(ℤ/5) (declared directly on
> MatrixElems(2, ℤ/5)), but the registered routes — element of
> Mat(_, ℚ) → sage "mat_det_q" — accept none of it.
> The method stays available on MatrixElems(2, ℤ/5); the
> mathematics is not narrowed — an implementation for this
> presentation does not exist yet.

```lean4
let N := [1, 2; 3, 4] in Mat₂(ℤ/5)
-- N.det()
```
